# Overfit and Underfit

```python
# %% [markdown]
# # 03_Overfit_Underfit — Overfitting and Underfitting (Keras 3 / TF2)
#
# You have already learned about overfitting and underfitting in classical machine learning. The same problems show up when training neural networks:
#
# - **Overfitting**: training error keeps improving, but validation error stagnates or gets worse.
# - **Underfitting**: both training and validation performance are poor (model too simple, too much regularization, or trained too briefly).
#
# Neural networks can overfit easily because they have high capacity. The best prevention is **better data coverage** (more diverse training data), but we also use **regularization**:
#
# - Reduce model capacity
# - L2 regularization (weight decay)
# - Dropout
#
# **Important (Keras 3 fix):**
# - Do **not** use `tf.keras.optimizers.legacy.*`
# - In Keras 3, **don’t reuse the same optimizer instance across different models**. Create a **fresh optimizer** per model.

# %%
# Imports
import os
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")

import tensorflow as tf

# Optional plotting helpers (safe to skip if not installed)
try:
    import tensorflow_docs as tfdocs
    import tensorflow_docs.plots
    TF_DOCS_AVAILABLE = True
except Exception:
    TF_DOCS_AVAILABLE = False

print("TensorFlow:", tf.__version__)
print("Eager:", tf.executing_eagerly())

# %% [markdown]
# ## Setup / Reproducibility

# %%
RSEED = 42
np.random.seed(RSEED)
tf.random.set_seed(RSEED)

# NOTE: do NOT force float64; it slows training a lot and is not needed here.
# tf.keras.backend.set_floatx("float64")

plt.rcParams["figure.figsize"] = (12, 7)

# %% [markdown]
# ## Load the Boston Housing dataset
#
# We will:
# 1. Load data from `../data/boston.csv`
# 2. Split into train/test
# 3. Scale numeric columns (fit scaler on train only)
# 4. Convert to NumPy arrays for Keras

# %%
df = pd.read_csv("../data/boston.csv")
display(df.head(3))
print("Shape:", df.shape)

# %%
print("Missing values per column:\n", df.isnull().sum())

# %%
# Target
target = df.pop("MEDV")

# Train/test split
X_train_df, X_test_df, y_train_s, y_test_s = train_test_split(
    df, target, random_state=RSEED
)

# Scale selected numeric columns (avoid leakage: fit on train only)
col_scale = ["CRIM","ZN","INDUS","NOX","RM","AGE","DIS","TAX","PTRATIO","LSTAT"]

scaler = StandardScaler()
X_train_df[col_scale] = scaler.fit_transform(X_train_df[col_scale])
X_test_df[col_scale]  = scaler.transform(X_test_df[col_scale])

# Convert to NumPy
X_train = X_train_df.values
X_test  = X_test_df.values
y_train = y_train_s.values
y_test  = y_test_s.values

print("X_train:", X_train.shape, "X_test:", X_test.shape)

# %% [markdown]
# ## Training utilities (Keras 3-safe)
#
# - We use a learning-rate schedule (optional but fine).
# - **Keras 3 fix:** create a **fresh optimizer per model**.
# - Use EarlyStopping to avoid wasting time (especially for huge models).

# %%
training_history = {}

EPOCHS = 2000
BATCH_SIZE = 32

N_TRAIN = X_train.shape[0]
steps_per_epoch = int(np.ceil(N_TRAIN / BATCH_SIZE))

lr_schedule = tf.keras.optimizers.schedules.InverseTimeDecay(
    initial_learning_rate=0.01,
    decay_steps=steps_per_epoch * 1000,
    decay_rate=1.0,
    staircase=False,
)

def compile_with_fresh_optimizer(model: tf.keras.Model) -> tf.keras.Model:
    """
    Keras 3: do NOT reuse optimizer objects across different models.
    Create a fresh optimizer instance each time you compile a new model.
    """
    opt = tf.keras.optimizers.Adam(learning_rate=lr_schedule, name="Adam")
    model.compile(optimizer=opt, loss="mae", metrics=["mse"])
    return model

early_stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=50,
    restore_best_weights=True,
)

def plot_metric(history, metric="mse", title=None):
    plt.plot(history.history[metric])
    plt.plot(history.history[f"val_{metric}"])
    plt.title(title or f"Model {metric.upper()}")
    plt.ylabel(metric.upper())
    plt.xlabel("Epoch")
    plt.legend(["train", "validation"], loc="upper right")
    plt.grid(True)
    plt.show()

def plot_loss(history, title="Model Loss"):
    plt.plot(history.history["loss"], label="loss")
    plt.plot(history.history["val_loss"], label="val_loss")
    plt.title(title)
    plt.xlabel("Epoch")
    plt.ylabel("MAE")
    plt.legend()
    plt.grid(True)
    plt.show()

# %% [markdown]
# ## Model architectures (builders)
#
# **Rule:** These functions should ONLY build the network (no `.compile()` inside).
# We compile later with a fresh optimizer per model.

# %%
n_features = X_train.shape[1]

def build_small_model():
    return tf.keras.Sequential([
        tf.keras.layers.Dense(32, kernel_initializer="uniform", activation="relu", input_shape=(n_features,)),
        tf.keras.layers.Dense(32, kernel_initializer="uniform", activation="relu"),
        tf.keras.layers.Dense(1, kernel_initializer="uniform"),
    ])

def build_medium_model():
    return tf.keras.Sequential([
        tf.keras.layers.Dense(64, kernel_initializer="uniform", activation="relu", input_shape=(n_features,)),
        tf.keras.layers.Dense(64, kernel_initializer="uniform", activation="relu"),
        tf.keras.layers.Dense(64, kernel_initializer="uniform", activation="relu"),
        tf.keras.layers.Dense(1, kernel_initializer="uniform"),
    ])

def build_large_model():
    return tf.keras.Sequential([
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu", input_shape=(n_features,)),
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu"),
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu"),
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu"),
        tf.keras.layers.Dense(1, kernel_initializer="uniform"),
    ])

def build_l2_model(l2=0.01):
    reg = tf.keras.regularizers.l2(l2)
    return tf.keras.Sequential([
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu", kernel_regularizer=reg, input_shape=(n_features,)),
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu", kernel_regularizer=reg),
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu", kernel_regularizer=reg),
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu", kernel_regularizer=reg),
        tf.keras.layers.Dense(1, kernel_initializer="uniform"),
    ])

def build_dropout_model(rate=0.25):
    return tf.keras.Sequential([
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu", input_shape=(n_features,)),
        tf.keras.layers.Dropout(rate),
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu"),
        tf.keras.layers.Dropout(rate),
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu"),
        tf.keras.layers.Dropout(rate),
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu"),
        tf.keras.layers.Dropout(rate),
        tf.keras.layers.Dense(1, kernel_initializer="uniform"),
    ])

def build_l2_dropout_model(l2=0.01, rate=0.25):
    reg = tf.keras.regularizers.l2(l2)
    return tf.keras.Sequential([
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu", kernel_regularizer=reg, input_shape=(n_features,)),
        tf.keras.layers.Dropout(rate),
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu", kernel_regularizer=reg),
        tf.keras.layers.Dropout(rate),
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu", kernel_regularizer=reg),
        tf.keras.layers.Dropout(rate),
        tf.keras.layers.Dense(512, kernel_initializer="uniform", activation="relu", kernel_regularizer=reg),
        tf.keras.layers.Dropout(rate),
        tf.keras.layers.Dense(1, kernel_initializer="uniform"),
    ])

# %% [markdown]
# ## Train: Small model

# %%
with tf.device("/cpu:0"):
    small_model = compile_with_fresh_optimizer(build_small_model())
    small_model.summary()

with tf.device("/cpu:0"):
    training_history["small"] = small_model.fit(
        X_train, y_train,
        validation_split=0.2,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=0,
        callbacks=[early_stop],
    )

plot_metric(training_history["small"], "mse", "Small model — MSE")
plot_loss(training_history["small"], "Small model — MAE loss")

# %%
loss, mse = small_model.evaluate(X_test, y_test, verbose=0)
print(f"Small model test MSE: {mse:.4f}")

y_pred = small_model.predict(X_test, verbose=0)
print("Sklearn MSE:", mean_squared_error(y_test, y_pred))

# %% [markdown]
# ## Train: Medium model
#
# This increases capacity and often reduces underfitting, but can overfit sooner.

# %%
with tf.device("/cpu:0"):
    medium_model = compile_with_fresh_optimizer(build_medium_model())
    medium_model.summary()

with tf.device("/cpu:0"):
    training_history["medium"] = medium_model.fit(
        X_train, y_train,
        validation_split=0.2,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=0,
        callbacks=[early_stop],
    )

plot_metric(training_history["medium"], "mse", "Medium model — MSE")
plot_loss(training_history["medium"], "Medium model — MAE loss")

# %%
loss, mse = medium_model.evaluate(X_test, y_test, verbose=0)
print(f"Medium model test MSE: {mse:.4f}")

# %% [markdown]
# ## Train: Large model
#
# This model has very high capacity and can memorize the training set.
# Expect stronger overfitting unless regularized.

# %%
with tf.device("/cpu:0"):
    large_model = compile_with_fresh_optimizer(build_large_model())
    large_model.summary()

with tf.device("/cpu:0"):
    training_history["large"] = large_model.fit(
        X_train, y_train,
        validation_split=0.2,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=0,
        callbacks=[early_stop],
    )

plot_metric(training_history["large"], "mse", "Large model — MSE")
plot_loss(training_history["large"], "Large model — MAE loss")

# %%
loss, mse = large_model.evaluate(X_test, y_test, verbose=0)
print(f"Large model test MSE: {mse:.4f}")

# %% [markdown]
# ## Regularization: L2 (weight decay)
#
# L2 discourages large weights and typically improves generalization for high-capacity networks.

# %%
with tf.device("/cpu:0"):
    l2_model = compile_with_fresh_optimizer(build_l2_model(l2=0.01))
    l2_model.summary()

with tf.device("/cpu:0"):
    training_history["l2"] = l2_model.fit(
        X_train, y_train,
        validation_split=0.2,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=0,
        callbacks=[early_stop],
    )

plot_metric(training_history["l2"], "mse", "Large + L2 — MSE")
plot_loss(training_history["l2"], "Large + L2 — MAE loss")

# %%
loss, mse = l2_model.evaluate(X_test, y_test, verbose=0)
print(f"L2 model test MSE: {mse:.4f}")

# %% [markdown]
# ## Regularization: Dropout
#
# Dropout randomly zeros activations during training to reduce co-adaptation.

# %%
with tf.device("/cpu:0"):
    dropout_model = compile_with_fresh_optimizer(build_dropout_model(rate=0.25))
    dropout_model.summary()

with tf.device("/cpu:0"):
    training_history["dropout"] = dropout_model.fit(
        X_train, y_train,
        validation_split=0.2,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=0,
        callbacks=[early_stop],
    )

plot_metric(training_history["dropout"], "mse", "Large + Dropout — MSE")
plot_loss(training_history["dropout"], "Large + Dropout — MAE loss")

# %%
loss, mse = dropout_model.evaluate(X_test, y_test, verbose=0)
print(f"Dropout model test MSE: {mse:.4f}")

# %% [markdown]
# ## Combine L2 + Dropout
#
# Common and often strong: L2 keeps weights small; dropout improves robustness.

# %%
with tf.device("/cpu:0"):
    l2_dropout_model = compile_with_fresh_optimizer(build_l2_dropout_model(l2=0.01, rate=0.25))
    l2_dropout_model.summary()

with tf.device("/cpu:0"):
    training_history["l2_dropout"] = l2_dropout_model.fit(
        X_train, y_train,
        validation_split=0.2,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=0,
        callbacks=[early_stop],
    )

plot_metric(training_history["l2_dropout"], "mse", "Large + L2 + Dropout — MSE")
plot_loss(training_history["l2_dropout"], "Large + L2 + Dropout — MAE loss")

# %%
loss, mse = l2_dropout_model.evaluate(X_test, y_test, verbose=0)
print(f"L2+Dropout model test MSE: {mse:.4f}")

# %% [markdown]
# ## Compare all models
#
# If `tensorflow_docs` is available, we can overlay all histories easily.
# Otherwise, this section is skipped.

# %%
if TF_DOCS_AVAILABLE:
    plotter = tfdocs.plots.HistoryPlotter(metric="mse", smoothing_std=10)
    plotter.plot(training_history)
    plt.xlim([5, max(plt.xlim())])
    plt.ylim([0, 50])
    plt.xlabel("Epochs")
    plt.title("Comparison of all models (MSE)")
    plt.legend(loc="upper right")
    plt.show()
else:
    print("tensorflow_docs not installed; skipping HistoryPlotter overlay.")

# %% [markdown]
# ## Summary
#
# Typical ways to prevent overfitting in neural networks:
#
# - Get more (and more diverse) training data
# - Reduce capacity (smaller network)
# - Add weight regularization (L2)
# - Add dropout
#
# Not covered here but important:
# - Data augmentation (mostly for images/audio/text)
# - Batch normalization
#
# **Keras 3 key fix used in this notebook:**
# - Fresh optimizer instance per model (do not reuse the same optimizer object).

# %% [markdown]
# ## Saving a model (SavedModel)
#
# This saves architecture + weights + training config into a folder.

# %%
os.makedirs("saved_model", exist_ok=True)

# Save the (unregularized) large model example
large_model.save("saved_model/my_large_model")

print("Saved model to: saved_model/my_large_model")
```

Notebook 3 – 03_Overfit_Underfit (Day 2) — Concise Summary

Purpose
Demonstrate underfitting vs overfitting in neural networks and show practical strategies to control model complexity and improve generalization using TensorFlow/Keras.

⸻

Core Concepts

Overfitting
	•	Training loss keeps improving while validation loss worsens.
	•	Caused by excessive model capacity, long training, or weak regularization.
	•	Results in poor performance on unseen (test) data.

Underfitting
	•	Both training and validation performance are poor.
	•	Model is too simple, over-regularized, or insufficiently trained.
	•	Fails to learn relevant patterns.

⸻

Data & Setup
	•	Dataset: Boston Housing (regression, target = MEDV)
	•	Preprocessing:
	•	Train/test split with fixed seed
	•	StandardScaler fit on training data only (no leakage)
	•	Converted to NumPy for Keras
	•	Metric: MSE (with MAE loss)
	•	Training:
	•	2000 epochs
	•	Adam optimizer with InverseTimeDecay learning rate schedule
	•	Centralized training_history dict for comparison

⸻

Model Complexity Experiments

Small Model
	•	2 hidden layers × 32 units
	•	Often underfits or saturates early

Medium Model
	•	3 hidden layers × 64 units
	•	Better bias–variance balance
	•	Strong baseline

Large Model
	•	4 hidden layers × 512 units
	•	Quickly overfits
	•	Lowest training loss, unstable validation loss without regularization

⸻

Regularization Strategies

L2 Regularization (Weight Decay)
	•	Penalizes large weights
	•	Keeps parameters small and stable
	•	Large + L2 ≈ Medium model performance, but more robust

Dropout
	•	Randomly zeroes activations during training
	•	Prevents co-adaptation
	•	Slower convergence, better generalization
	•	Typical rates: 0.2–0.5 (used: 0.25)

L2 + Dropout (Combined)
	•	L2 controls weight magnitude
	•	Dropout adds noise & robustness
	•	Strongest overall generalization pattern

⸻

Key Results
	•	Unregularized large model: best train MSE, poor generalization
	•	Large + L2: competitive with medium model, far less overfitting
	•	Dropout & L2+Dropout: smoother validation curves, delayed overfitting
	•	Dataset itself limits dramatic overfitting, but patterns are clear

⸻

Practical Takeaways

To reduce overfitting in neural networks:
	•	Get more diverse data (not just more of the same)
	•	Reduce model capacity
	•	Use L2 regularization
	•	Use Dropout
	•	Combine techniques when needed

Not covered but important:
	•	Data augmentation
	•	Batch normalization

⸻

Model Persistence
	•	Demonstrated:
	•	Checkpointing (ModelCheckpoint)
	•	Full model saving via SavedModel
	•	Enables:
	•	Reloading without code
	•	Continued training
	•	Deployment (TF Serving, TF Lite, TF.js)

You have already learned about overfitting and underfitting in the context of machine learning algorithms. Unfortunately, these topics are also important when it comes to training an artificial neural network.

You may have observed in previous notebooks that the accuracy(/mae) of your model on the validation set reaches a peak(/minimum) after training for a number of epochs and then stagnates or, worse, starts to decrease(/increase). This is a clear sign that the model is overfitting the training data. 

Artificial neural networks tend to overfit the data. It is possible to achieve extremely high accuracy (or low mae) on your training data if you just train your model long enough. However, this is not desirable because these models do not generalize well and will perform poorly on a test set or new data.

Underfitting is another problem. If your model is not powerful enough, it is over-regularized, or it has not been trained long enough, it may perform poorly on the training and test set because it has not been able to learn the relevant features from the data.

The best way to prevent your model from overfitting is using a dataset for training that covers the full range of inputs that the model is expected to handle. Additional data may only be useful if it covers new and interesting cases.

However, sometimes you will need to use different regularization techniques to counteract the overfitting of the data. In this notebook we will explore some common techniques that can help you with this task.

## Strategies to prevent overfitting

### Regularization 

You may be familiar with Occam's Razor principle: given two explanations for something, the explanation most likely to be correct is the "simplest" one, the one that makes the least amount of assumptions. This also applies to the models learned by neural networks: given some training data and a network architecture, there are multiple sets of weights values (multiple models) that could explain the data, and simpler models are less likely to overfit than complex ones.

In the next section we will try out one form of **regularization**. Hopefully, you all are familiar with L1 and L2 regularization as we talked about it several times during the bootcamp. Nevertheless, let's refresh the general concepts before we move on: 

We can apply regularization techniques in order generate a simpler model to prevent overfitting. A "simple model" in this context is a model where the distribution of parameter values has less entropy (or a model with fewer parameters altogether). Thus a common way to mitigate overfitting is to put constraints on the complexity of a network by forcing its weights only to take small values, which makes the distribution of weight values more "regular". It is done by adding to the loss function of the network a cost associated with having large weights. This cost comes in two flavors:

* [L1 regularization](https://developers.google.com/machine-learning/glossary/#L1_regularization), where the cost added is proportional to the absolute value of the weights coefficients (i.e. to what is called the "L1 norm" of the weights).

* [L2 regularization](https://developers.google.com/machine-learning/glossary/#L2_regularization), where the cost added is proportional to the square of the value of the weights coefficients (i.e. to what is called the squared "L2 norm" of the weights). L2 regularization is also called weight decay in the context of neural networks. Don't let the different name confuse you: weight decay is mathematically the exact same as L2 regularization.

L1 regularization pushes weights towards exactly zero encouraging a sparse model. L2 regularization will penalize the weights parameters without making them sparse since the penalty goes to zero for small weights - one reason why L2 is more common.

In `tf.keras`, weight regularization is added by passing weight regularizer instances to layers as keyword arguments. Let's add L2 weight regularization to our large model from above:

`l2(0.01)` means that every coefficient in the weight matrix of the layer will add `0.01 * weight_coefficient_value**2` to the total **loss** of the network.

If you still see overfitting, increase the coefficient (for example to `l2(0.02)`). If training stalls, decrease it. The L2 term is applied per layer, so you can regularize only the dense layers that carry most of the parameters instead of every layer.

### Dropout

Dropout is one of the most effective and most commonly used regularization techniques for neural networks, developed by Hinton and his students at the University of Toronto.

The intuitive explanation for dropout is that because individual nodes in the network cannot rely on the output of the others, each node must output features that are useful on their own.

Dropout, applied to a layer, consists of randomly "dropping out" (i.e. set to zero) a number of output features of the layer during training. Let's say a given layer would normally have returned a vector [0.2, 0.5, 1.3, 0.8, 1.1] for a given input sample during training; after applying dropout, this vector will have a few zero entries distributed at random, e.g. [0, 0.5, 1.3, 0, 1.1].

The "dropout rate" is the fraction of the features that are being zeroed-out; it is usually set between 0.2 and 0.5. At test time, no units are dropped out, and instead the layer's output values are scaled down by a factor equal to the dropout rate, so as to balance for the fact that more units are active than at training time.

In `tf.keras` you can introduce dropout in a network via the Dropout layer, which gets applied to the output of layer right before.

Let's add a Dropout layer after each dense layer in our network to see how they influence our models performance.

Dropout usually slows convergence a bit because the network sees a noisier view of itself each batch. That's expected—look for validation curves that are smoother and peak later, indicating better generalization. You can sweep rates like 0.2, 0.3, 0.5 to see the trade-off between regularization strength and training speed.

You have already learned about overfitting and underfitting in the context of machine learning algorithms. Unfortunately, these topics are also important when it comes to training an artificial neural network.

You may have observed in previous notebooks that the accuracy(/mae) of your model on the validation set reaches a peak(/minimum) after training for a number of epochs and then stagnates or, worse, starts to decrease(/increase). This is a clear sign that the model is overfitting the training data. 

Artificial neural networks tend to overfit the data. It is possible to achieve extremely high accuracy (or low mae) on your training data if you just train your model long enough. However, this is not desirable because these models do not generalize well and will perform poorly on a test set or new data.

Underfitting is another problem. If your model is not powerful enough, it is over-regularized, or it has not been trained long enough, it may perform poorly on the training and test set because it has not been able to learn the relevant features from the data.

The best way to prevent your model from overfitting is using a dataset for training that covers the full range of inputs that the model is expected to handle. Additional data may only be useful if it covers new and interesting cases.

However, sometimes you will need to use different regularization techniques to counteract the overfitting of the data. In this notebook we will explore some common techniques that can help you with this task.
